# Array Slicing

___

One important concept we expect you to be comfortable with coming into this course is array slicing. This is really important as it will be used every lab.

Recall that if 'arr' is an array, then arr[START :STOP] gives the elements of the array from START to STOP.

Please complete the following questions to review this concept. If you're unfamiliar with array slicing, come back to this question once you've reviewed the relevant material in the following notebooks.

First, define a numpy array `int_array`. The array is provided in the Prairelearn question page.

```python
#grade (enter your code in this cell - DO NOT DELETE THIS LINE) 
int_array=np.array([5, 0, 19, 8, 18,  5 ])
```

Now, using **only array slicing**, create a new list `middle` which contains all of the entries of `int_array` expect the first and last entries.

Example: If `int_array`=[0, 1, 2, 17, 5], then `middle`=[1, 2, 17].

```python
#grade (enter your code in this cell - DO NOT DELETE THIS LINE) 
middle = np.array(int_array[1:5])
```

Next, using **only array slicing**, create a list `odds` which contains only the odd-indexed entries of `int_array`.

Example: If `int_array`=[0, 1, 2, 34, 17, 5], then `odds`=[1, 34, 5].

```python
#grade (enter your code in this cell - DO NOT DELETE THIS LINE) 
odds = np.array(int_array[1:6:2])
```
